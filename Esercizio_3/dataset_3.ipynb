```python
from wordfreq import top_n_list
import pandas as pd
import random
import difflib
from wordfreq import zipf_frequency
```

```python
#filtro per ottenere solo parole alfabetiche
vocab = [
    w for w in top_n_list("en", 120000)
    if w.isalpha()
    if len(w)  >= 3
]
```

```python
#prendo le parole più frequenti della lingua inglese (30k,50k,100k)
vocab_30k = vocab[:30000]
vocab_50k = vocab[:50000]
vocab_100k = vocab[:100000]
print(len(vocab_100k))
```

```python
#si salva vocab 100k serve per dopo
with open("vocab_100k.txt", "w", encoding="utf-8") as f:
    for word in vocab_100k:
        f.write(word + "\n")
```

```python
#unisco le liste per controllo
all = (
    Sing_plu
    + Genere
    + Cap_paes
    + Agg_avv
)
```

```python
#controllo che le parole ci siano nei vocabolari
all_words = set()

for a, b in all:
    all_words.update([a, b])

print("Mancanti nel 30k:")
print(sorted(all_words - set(vocab_30k)))

print("\nMancanti nel 50k:")
print(sorted(all_words - set(vocab_50k)))

print("\nMancanti nel 100k:")
print(sorted(all_words - set(vocab_100k)))
```

```python
#creo le analogie
analogies = []

for category, pairs in [
    ("plurale", Sing_plu),
    ("genere", Genere),
    ("capitale_paese", Cap_paes),
    ("avverbi", Agg_avv)
]:
    for i in range(35):
        a, b = pairs[i]
        c, d = pairs[(i + 1) % 10]

        analogies.append({
            "category": category,
            "a": a,
            "b": b,
            "c": c,
            "d": d
        })
        
        
print(len(analogies))
```

```python
#creiamo dataframe
df_analogie = pd.DataFrame(analogies)

print(df_analogie.head())

df_analogie.to_csv(
    "Analogie.csv",
    index=False
)

df_analogie["category"].value_counts()
```
